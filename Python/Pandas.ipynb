5. Grouping Data

select country,
sum(amount)
from transactions
group by country

```python
import pandas as pd

country = {
    
    "countryname":['USA',"UK","UEA","INDIA",'USA','UK',"UEA","INDIA"],
    "amount":[200,500,1000,200,500,1000,200,500],
    "fraud":[0,1,0,1,0,1,0,1]
}

#Single aggregation for the above DataFrame:
country = pd.DataFrame(country)
print(country.groupby('countryname')['amount'].sum()) # Group the DataFrame by 'countryname' and calculate the sum of 'amount' for each group 

#Multiple aggregations:
print(country.groupby('countryname')['amount'].agg(['sum', 'mean', 'max', 'min'])) # Group the DataFrame by 'countryname' and calculate the sum, mean, max, and min of 'amount' for each group

#rename new columns with aggregation functions
country.groupby('countryname')['amount'].agg(['sum', 'mean', 'max', 'min']).rename(columns={'sum': 'total_amount', 'mean': 'average_amount', 'max': 'max_amount', 'min': 'min_amount'})
```

6. Data Selection from Dataframe    

```python
# Select specific columns from the DataFrame
print(file_df['Amount'])# Print the 'Amount' column from the DataFrame
print(file_df[['Amount','Time','V2']]) # Print multiple columns from the dataframe.

# Filter rows based on a condition
print(file_df[file_df['Amount']>1000])  # Filter and print rows where the 'Amount' column is greater than 1000
print(file_df[file_df['Class']==1])  # Filter and print rows where the 'Class' column is equal to 1
print(file_df[(file_df['Amount']>1000) & (file_df['Class']==1)])  # Filter and print rows where the 'Amount' column is greater than 1000 and the 'Class' column is equal to 1
#select specific rows using iloc
print(file_df.iloc[0:10]) # Select and print rows from integer based index 0 to 09 using iloc based on integer-based indexing
print(file_df.loc[0:10]) # Select and print rows from label based index 0 to 10 using loc based on label-based indexing

#Data Sorting
print(file_df.sort_values(by='Amount', ascending=False)) # Sort the DataFrame by the 'Amount' column in descending order
print(file_df.sort_values(by=['Amount', 'Time'], ascending=[False, True])) # Sort the DataFrame by 'Amount' in descending order and 'Time' in ascending order
print(file_df.sort_index(ascending=True)) # Sort the DataFrame by index in ascending order

#Change Datatypes
print(file_df.dtypes) # Display the data types of each column in the DataFrame
print(file_df['V2'].astype(int)) # Change the data type of the 'Transaction_Amount' column to float

#rename columns
print(file_df.rename(columns={'Amount' : 'Transaction_Amount', 'Time' : 'Transaction_Time'})) # Rename the 'Amount' column to 'Transaction_Amount' and the 'Time' column to 'Transaction_Time'
print(file_df.rename(columns={'Amount' : 'Transaction_Amount', 'Time' : 'Transaction_Time'}, inplace=True)) # Rename the 'Amount' column to 'Transaction_Amount' and the 'Time' column to 'Transaction_Time' and modify the original DataFrame in place

```

```python

# Difference between loc and iloc:
# loc is label-based indexing, which means that you have to specify the name of the rows    
# iloc is integer-based indexing, which means that you have to specify the index of the rows    
# syntax for loc is df.loc[row_labels, column_labels]
# syntax for iloc is df.iloc[row_indexer, column_indexer]

print(file_df.loc[file_df['Transaction_Amount'] > 1000, ['Transaction_Amount', 'Transaction_Time']].head(10)) 
# Select and print the 'Amount' and 'Time' columns for rows where the 'Amount' column is greater than 1000
print(file_df.iloc[file_df['Transaction_Amount'] > 1000,[0,2]])
 # Select and print the 'Amount' and 'Time' columns for rows where the 'Amount' column is greater than 1000 and convert the result to a dictionary

print(file_df.loc[file_df['Transaction_Amount'] > 1000, 'Transaction_Amount'] > 50)
# Select and print the Amount and Time columns for rows where the Amount column is greater than 1000 and check if the result is greater than 50
```
