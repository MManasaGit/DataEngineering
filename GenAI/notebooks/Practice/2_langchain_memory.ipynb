```python
from langchain_openai import ChatOpenAI
import dotenv
import os
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

load_env = dotenv.load_dotenv()
```

```python
llm = ChatOpenAI(model="gpt-4o", temperature=0.7, max_tokens = 1000,
                model_kwargs = {"api_key": os.environ.get("OPENAI_API_KEY")})

prompt = "What is the capital of France"
invoke_llm = llm.invoke(prompt)
print (invoke_llm.content)
```

```python
prompt = "Which place did i refer to in the conversation?"
invoke_llm = llm.invoke(prompt) 
print(invoke_llm.content)
```

#### Above code will give answer to your prompt but will not remember the history

```python
# conversation memory is not maintained across different invocations of the llm.invoke() method. Each invocation is stateless and does not retain any context from previous calls. To maintain conversation memory, you would need to implement a mechanism to store and pass the conversation history to the model in each prompt.


conversation_history = [
    ("user", "What is the capital of France?\n"),
    ("ai", "The capital of France is Paris.\n"),
    ("user", "Which place did I refer to in the conversation?\n")
]
# Conversations can be stored in the form of a list of tuples, where each tuple contains the speaker (user or AI) and their respective message. This allows for easy tracking of the conversation flow and context.
# alternative way to store above conversational history is to use a list of dictionaries, where each dictionary represents a message in the conversation. Each dictionary can have keys for the speaker (role) and the message content. This structure allows for more flexibility and easier access to specific parts of the conversation.

conversation_history = [
    {"role": "user", "content": "What is the capital of France?\n"},
    {"role": "ai", "content": "The capital of France is Paris.\n"},
    {"role": "user", "content": "Which place did I refer to in the conversation?\n"}
]

# Below is an example of how to use the conversation history to maintain context in a conversation with the model. The conversation history is passed as input to the model, allowing it to generate responses based on the previous messages in the conversation.
conversation_history = [
    HumanMessage(content="What is the capital of France?"),
    AIMessage(content="The capital of France is Paris."),
    HumanMessage(content="Which place did I refer to in the conversation?")
]
invoke_llm = llm.invoke(conversation_history)
print(invoke_llm.content)

```

#### Above code will now remember the history and give the relevent answers. However, This is static way of working with LLMs

```python
conversation_history = []

while True:
    user_input = input("User: ")
    conversation_history.append(("user", user_input + "\n"))
    if user_input.lower() in ["exit", "quit", "bye"]:
        print("AI:","Bye!")
        break

    invoke_llm = llm.invoke(conversation_history)
    print("AI:", invoke_llm.content)

    conversation_history.append(("ai", invoke_llm.content + "\n"))
```

#### The above code will Dynamically take the each input as a conversation history gets the answer per session. 

#### If the user exists then the session will end and we re-execute to get into a new session.
