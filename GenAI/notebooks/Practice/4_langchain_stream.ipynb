```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(model_name="gpt-4o", temperature= 0.8,streaming=True)
llm_response = llm.stream("Get me the name, age and qulification of the top 5 sports personalities in the world")
for chunk in llm_response:
    print(chunk.content, end="")
```
