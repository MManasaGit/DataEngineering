### Prompt template or Dynamic prompts

```python
from langchain_core.prompts import ChatPromptTemplate

llmprompt = ChatPromptTemplate.from_messages([
    ("system", "You are a translator translate between English and {language}."),
    ("user", "{text}")]
)

llmprompt
```

##### Pass the parameters using format_message

```python
pass_args = llmprompt.format_messages(language="Portuguese", text="My name is Manasa. I am a Data engineer.")
pass_args#[1].content
```

#### Invoke the llm with the dynamic prompt message

```python
llminvoke = llm.invoke(pass_args)
llminvoke.content
```

#### Chains - runnable

```python
llmout = llm.invoke(llmprompt.invoke({"language": "Portuguese", "text": "My name is Manasa. I am a Data engineer."})) 
# Here we are using the llmprompt to format the prompt and then passing it to the llm.invoke method to get the response.
llmout.content
# explained in mid of https://www.youtube.com/watch?v=GFZcqo77xIs
```

#### CHAINS

```python
Chains = llmprompt | llm 
# Chains are a way to combine multiple components together to create a single pipeline. In this case, we are combining the llmprompt and llm together to create a single pipeline that takes in a language and text and returns the translated text.

Chains.invoke({"language": "Portuguese", "text": "My name is Manasa. I am a Data engineer."}).content
```

### Now we saw 3 ways to invoke the llm with the prompt template. 
##### The first way is to use the llmprompt.format_messages method to format the prompt and then pass it to the llm.invoke method. 
##### The second way is to use the llmprompt.invoke method to format the prompt and then pass it to the llm.invoke method.
##### The third way is to use the Chains class to combine the llmprompt and llm together and then invoke the chain with the input parameters.

#### Chains for output

```python
from langchain_core.output_parsers import  StrOutputParser

out = StrOutputParser()
Chains = llmprompt | llm | out
llminvoke = Chains.invoke({"language": "Portuguese", "text": "My name is Manasa. I am a Data engineer."})
llminvoke
# here we are using the StrOutputParser to parse the output of the llm.invoke method and return a string. This is useful when we want to get the output of the llm.invoke method as a string instead of a list of messages.
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.prompts import BaseChatPromptTemplate
from langchain_core.prompts import FewShotPromptTemplate
from langchain_core.prompts import FewShotChatMessagePromptTemplate

# ============================================================
# 1. PromptTemplate
# ============================================================


message = PromptTemplate.from_template(
    "Explain {topic} in real terms"
)

prompts = message.invoke("RAG")

print(prompts)

# ============================================================
# 2. ChatPromptTemplate
# ============================================================

message = ChatPromptTemplate.from_messages([
    ("system","You are helpful AI agent"),
    ("human","What is {RAG}"),
    ("ai", "I can help you!!")
])

prompts = message.invoke("invoke")

print(prompts)

# ============================================================
# 3. ChatPromptTemplate
# ============================================================

examples = [
    {
        "input": "I love this product.",
        "output": "Positive"
    },
    {
        "input": "This product is terrible.",
        "output": "Negative"
    }
]

example_prompt = ChatPromptTemplate.from_messages([
    ("human", "{input}"),
    ("ai", "{output}")
])

few_shot = FewShotChatMessagePromptTemplate(
    examples=examples,
    example_prompt=example_prompt
)

prompt = ChatPromptTemplate.from_messages([
    ("system", "Classify the sentiment."),
    few_shot,
    ("human", "{input}")
])

messages = prompt.invoke({
    "input": "I really like this product."
})


for message in messages.messages:
    print(message.type, ":", message.content)

# -----------------------------------------
# 7. Send messages to LLM
# -----------------------------------------

from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model_name="gpt-4", temperature=0.9)

response = llm.invoke(messages)

print("ai:",response.content)
```
