```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain.agents import create_agent
from langchain_community import OpenAICommunity
```

```python
@tool
def sum_numbers(a: int, b: int) -> int:
    """Returns the sum of two numbers."""
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """Returns the product of two numbers."""
    return a * b
```

```python
llm = ChatOpenAI(model_name="gpt-4o", temperature= 0.8,streaming=True)
prompt = {"messages": [{"role": "user", "content": "What is the sum of 5 and 10?"}]}
agent = create_agent(llm, tools=[sum_numbers, multiply_numbers])
agent_response = agent.invoke(prompt)
for response in agent_response['messages']:
    print(response.content, end="\n")
```
